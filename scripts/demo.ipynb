```python
from atomx import Atom, Correlation

# Neutral silver with the default Dirac-Fock SCF: shooting engine, HF
# exchange, Breit + QED on, Guerra CI averaging for the open 5s shell.
atom = Atom(47, engine="shooting", exchange="hf",
            correlation=Correlation(averaging="ci"))
atom.solve()
print(atom.total_energy)
print(atom.orbital_energies)
```
