```python
from atomx import Atom

# Neutral cobalt with default Dirac-Fock SCF
atom = Atom(Z=82,qed=True, breit=True, exchange="hf", numerics="shooting")
atom.solve()
print(atom.total_energy)       # -128.04 Hartree
print(atom.orbital_energies)   # {1s: -32.6, 2s: -1.6, ...}
```
