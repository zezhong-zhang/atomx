```python
from atomx import Atom, Correlation

# Neutral neon with the default Dirac-Fock SCF: shooting engine, HF
# exchange, Breit + QED on, Guerra CI averaging for the open 2p shell.
atom = Atom(10, engine="shooting", exchange="hf",
            correlation=Correlation(averaging="ci"))
atom.solve()
print(atom.total_energy)       # -128.63 Hartree
print(atom.orbital_energies)   # {'1s1/2': -32.6, '2s1/2': -1.6, ...}
```
